# Operation Across Dimensions
this is so fundamental for pytorch. Obv simple operations can be done one 1 dimensionsal tensors

```python
t=torch.tensor([0.5,1,3,4])
torch.mean(t), torch.std(t), torch.max(t), torch.min(t)
```

but suppsose we have 2d tensor ,for ex and want to compute the mean value fo each columns

```python
t=torch.arange(20,dtype=float).reshape(5,4)
```

```python
t[0]
```

```python
print(t)
```

```python
torch.mean(t,axis=0).shape
```

this can be done for higher dimensionally arrays as well

```python
t=torch.randn(5,256,256,3)
```

```python
torch.mean(t,axis=0).shape
```

take the mean across the color channels

```python
torch.mean(t,axis=-1).shape
```

```python
x=torch.tensor([[5.,8.],[4.,6.]], requires_grad=True)
y=x.pow(3).sum()
y
x
```

```python
y.backward() #compute the gradient
x.grad # print the gradient
```

```python
x.grad
```

***Any  sort of large matrix multiplication problem is faster with torch tensors that it is with numpy arrays especially when you are running gpu***

```python
A=torch.randn((1000,1000))
B=torch.randn((1000,1000))
```

```python
t1=time.perf_counter()
torch.matmul(A,B)
t2=time.perf_counter()
print(t1-t2)
```
